```python
import pandas as pd

labels = pd.read_csv('../data/raw/SCIN/scin_labels.csv')
print(labels.shape)
print(labels.columns.tolist())
labels.head()
```

```python
# Define your target conditions
target_conditions = ['Vitiligo', 'Melasma', 'Psoriasis', 'Eczema', 'Tinea', 'Dermatitis']

# Search for them in the labels column
condition_counts = {}

for condition in target_conditions:
    count = labels['dermatologist_skin_condition_on_label_name'].astype(str).str.contains(condition, case=False).sum()
    condition_counts[condition] = count

for condition, count in condition_counts.items():
    print(f"{condition}: {count} images")
```

DermaCon-IN

```python
import pandas as pd

derma = pd.read_csv('../data/raw/DermaCon-IN/METADATA/Skin_Metadata-1.csv')

target_conditions = ['Vitiligo', 'Melasma', 'Psoriasis', 'Eczema', 'Tinea', 'Dermatitis']

for condition in target_conditions:
    count = derma['Disease_label'].astype(str).str.contains(condition, case=False).sum()
    print(f"{condition}: {count} images")
```

```python
print(derma['Main_class'].unique())
```

```python
print(derma['Sub_class'].unique())
```

```python
# Get all disease labels that match our target conditions
for condition in target_conditions:
    matches = derma[derma['Disease_label'].astype(str).str.contains(condition, case=False)]['Disease_label'].value_counts()
    print(f"\n{condition}:")
    print(matches)
```

```python
contact = derma[derma['Disease_label'].astype(str).str.contains('Contact Dermatitis', case=False)]['Disease_label'].value_counts()
seborrheic = derma[derma['Disease_label'].astype(str).str.contains('Seborrheic', case=False)]['Disease_label'].value_counts()

print("Contact Dermatitis variants:")
print(contact)
print("\nSeborrheic Dermatitis variants:")
print(seborrheic)
```
